```python
import requests

def get_token_data(symbol):
    url = "https://api.coingecko.com/api/v3/coins/markets"
    params = {
        "vs_currency": "usd",
        "ids": symbol.lower(), 
        "order": "market_cap_desc",
        "per_page": 1,
        "page": 1,
        "sparkline": False
    }

    response = requests.get(url, params=params)
    
    if response.status_code == 200:
        data = response.json()
        if data:
            token = data[0]
            return {
                "Symbol": token["symbol"].upper(),
                "Name": token["name"],
                "Price (USD)": token["current_price"],
                "24h Change (%)": token["price_change_percentage_24h"],
                "Market Cap": token["market_cap"],
                "Last Updated": token["last_updated"]
            }
        else:
            return {"Error": "Token not found"}
    else:
        return {"Error": "API request failed"}

# Example: Fetch data for SushiSwap token
token_symbol = "uniswap"
token_data = get_token_data(token_symbol)

print(token_data)
```
